```python
from sklearn.model_selection import train_test_split

# Get image filenames
image_files = sorted(os.listdir(train_image_dir))

# Keep only image files
image_files = [
    f for f in image_files
    if f.lower().endswith((".jpg", ".jpeg", ".png"))
]

# Split
train_files, val_files = train_test_split(
    image_files,
    test_size=0.20,
    random_state=42,
    shuffle=True
)

print("Training samples   :", len(train_files))
print("Validation samples :", len(val_files))
```

```python
from torch.utils.data import Dataset
import torch

class KvasirDataset(Dataset):

    def __init__(self, image_files, image_dir, mask_dir, image_size=256):
        self.image_files = image_files
        self.image_dir = image_dir
        self.mask_dir = mask_dir
        self.image_size = image_size

    def __len__(self):
        return len(self.image_files)

    def __getitem__(self, idx):

        # -------------------------
        # Image filename
        # -------------------------
        image_name = self.image_files[idx]

        # -------------------------
        # Load image
        # -------------------------
        image_path = os.path.join(
            self.image_dir,
            image_name
        )

        image = cv2.imread(image_path)

        if image is None:
            raise FileNotFoundError(
                f"Could not read image: {image_path}"
            )

        image = cv2.cvtColor(
            image,
            cv2.COLOR_BGR2RGB
        )

        # -------------------------
        # Corresponding mask
        # -------------------------
        mask_name = os.path.splitext(image_name)[0] + ".jpg"

        mask_path = os.path.join(
            self.mask_dir,
            mask_name
        )

        mask = cv2.imread(
            mask_path,
            cv2.IMREAD_GRAYSCALE
        )

        if mask is None:
            raise FileNotFoundError(
                f"Could not read mask: {mask_path}"
            )

        # -------------------------
        # Resize
        # -------------------------
        image = cv2.resize(
            image,
            (self.image_size, self.image_size)
        )

        mask = cv2.resize(
            mask,
            (self.image_size, self.image_size),
            interpolation=cv2.INTER_NEAREST
        )

        # -------------------------
        # Normalize image
        # -------------------------
        image = image.astype(np.float32) / 255.0

        # -------------------------
        # Binary mask
        # -------------------------
        mask = (mask > 0).astype(np.float32)

        # -------------------------
        # HWC → CHW
        # -------------------------
        image = np.transpose(
            image,
            (2, 0, 1)
        )

        # H,W → 1,H,W
        mask = np.expand_dims(
            mask,
            axis=0
        )

        # -------------------------
        # Convert to tensors
        # -------------------------
        image = torch.tensor(
            image,
            dtype=torch.float32
        )

        mask = torch.tensor(
            mask,
            dtype=torch.float32
        )

        return image, mask
```

```python
class DoubleConv(nn.Module):

    def __init__(self, in_channels, out_channels):
        super().__init__()

        self.conv = nn.Sequential(
            nn.Conv2d(in_channels, out_channels, kernel_size=3, padding=1),
            nn.ReLU(inplace=True),

            nn.Conv2d(out_channels, out_channels, kernel_size=3, padding=1),
            nn.ReLU(inplace=True)
        )

    def forward(self, x):
        return self.conv(x)
```

```python
import torch
import torch.nn as nn
```

```python
class UNet(nn.Module):

    def __init__(self):
        super().__init__()

        # Encoder
        self.enc1 = DoubleConv(3, 64)
        self.enc2 = DoubleConv(64, 128)
        self.enc3 = DoubleConv(128, 256)
        self.enc4 = DoubleConv(256, 512)

        # Pooling
        self.pool = nn.MaxPool2d(2)

        # Bottleneck
        self.bottleneck = DoubleConv(512, 1024)

        # Decoder
        self.up4 = nn.ConvTranspose2d(1024, 512, 2, stride=2)
        self.dec4 = DoubleConv(1024, 512)

        self.up3 = nn.ConvTranspose2d(512, 256, 2, stride=2)
        self.dec3 = DoubleConv(512, 256)

        self.up2 = nn.ConvTranspose2d(256, 128, 2, stride=2)
        self.dec2 = DoubleConv(256, 128)

        self.up1 = nn.ConvTranspose2d(128, 64, 2, stride=2)
        self.dec1 = DoubleConv(128, 64)

        # Final output
        self.out = nn.Conv2d(64, 1, kernel_size=1)

    def forward(self, x):

        # Encoder
        e1 = self.enc1(x)
        e2 = self.enc2(self.pool(e1))
        e3 = self.enc3(self.pool(e2))
        e4 = self.enc4(self.pool(e3))

        # Bottleneck
        b = self.bottleneck(self.pool(e4))

        # Decoder
        d4 = self.up4(b)
        d4 = torch.cat([d4, e4], dim=1)
        d4 = self.dec4(d4)

        d3 = self.up3(d4)
        d3 = torch.cat([d3, e3], dim=1)
        d3 = self.dec3(d3)

        d2 = self.up2(d3)
        d2 = torch.cat([d2, e2], dim=1)
        d2 = self.dec2(d2)

        d1 = self.up1(d2)
        d1 = torch.cat([d1, e1], dim=1)
        d1 = self.dec1(d1)

        return self.out(d1)
```

```python
import torch
import torch.nn as nn

# Binary Cross Entropy
bce_loss = nn.BCEWithLogitsLoss()


# Dice Loss
def dice_loss(pred, target, smooth=1.0):

    # Convert logits to probabilities
    pred = torch.sigmoid(pred)

    # Flatten
    pred = pred.view(-1)
    target = target.view(-1)

    # Dice coefficient
    intersection = (pred * target).sum()

    dice = (
        (2.0 * intersection + smooth)
        /
        (pred.sum() + target.sum() + smooth)
    )

    return 1 - dice


# Combined loss
def combined_loss(pred, target):

    bce = bce_loss(pred, target)
    dice = dice_loss(pred, target)

    return bce + dice
```

Create a Dice score for monitoring

```python
def dice_score(pred, target, smooth=1.0):

    # Convert logits to probability
    pred = torch.sigmoid(pred)

    # Convert probability to binary mask
    pred = (pred > 0.5).float()

    # Flatten
    pred = pred.view(-1)
    target = target.view(-1)

    intersection = (pred * target).sum()

    dice = (
        (2.0 * intersection + smooth)
        /
        (pred.sum() + target.sum() + smooth)
    )

    return dice.item()
```

```python
num_epochs = 20

train_losses = []
val_losses = []
val_dices = []

best_val_loss = float("inf")

for epoch in range(num_epochs):

    # =========================
    # TRAINING
    # =========================
    model.train()

    running_train_loss = 0.0

    for images, masks in train_loader:

        images = images.to(device)
        masks = masks.to(device)

        # Clear previous gradients
        optimizer.zero_grad()

        # Forward pass
        outputs = model(images)

        # Calculate loss
        loss = combined_loss(outputs, masks)

        # Backpropagation
        loss.backward()

        # Update weights
        optimizer.step()

        running_train_loss += loss.item()

    train_loss = (
        running_train_loss / len(train_loader)
    )

    # =========================
    # VALIDATION
    # =========================
    model.eval()

    running_val_loss = 0.0
    running_dice = 0.0

    with torch.no_grad():

        for images, masks in val_loader:

            images = images.to(device)
            masks = masks.to(device)

            # Prediction
            outputs = model(images)

            # Validation loss
            loss = combined_loss(
                outputs,
                masks
            )

            running_val_loss += loss.item()

            # Dice score
            running_dice += dice_score(
                outputs,
                masks
            )

    val_loss = (
        running_val_loss / len(val_loader)
    )

    val_dice = (
        running_dice / len(val_loader)
    )

    # Store results
    train_losses.append(train_loss)
    val_losses.append(val_loss)
    val_dices.append(val_dice)

    # Save best model
    if val_loss < best_val_loss:

        best_val_loss = val_loss

        torch.save(
            model.state_dict(),
            "/content/best_unet.pth"
        )

    print(
        f"Epoch [{epoch+1}/{num_epochs}] "
        f"Train Loss: {train_loss:.4f} | "
        f"Val Loss: {val_loss:.4f} | "
        f"Val Dice: {val_dice:.4f}"
    )
```

```python
def calculate_metrics(pred, target):

    pred = torch.sigmoid(pred)
    pred = (pred > 0.5).float()

    pred = pred.view(-1)
    target = target.view(-1)

    intersection = (pred * target).sum()

    dice = (
        (2 * intersection + 1e-7)
        /
        (pred.sum() + target.sum() + 1e-7)
    )

    union = (
        pred.sum()
        + target.sum()
        - intersection
    )

    iou = (
        (intersection + 1e-7)
        /
        (union + 1e-7)
    )

    return dice.item(), iou.item()
```

```python
test_dice = []
test_iou = []

with torch.no_grad():

    for images, masks in test_loader:

        images = images.to(device)
        masks = masks.to(device)

        outputs = model(images)

        for i in range(images.size(0)):

            dice, iou = calculate_metrics(
                outputs[i],
                masks[i]
            )

            test_dice.append(dice)
            test_iou.append(iou)

print("Final Test Dice:", sum(test_dice) / len(test_dice))
print("Final Test IoU :", sum(test_iou) / len(test_iou))
```
